```python
#COVID-19 Global Data Analysis
#project 2 | Medha Mishra
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load built-in dataset
df = sns.load_dataset("healthexp")

print("shape:", df.shape)
print(df.head(10))
```

```python
#step 3 - explore the data
print("countries:", df["Country"].unique())
print("\nyears:", df["Year"].min(), "to", df["Year"].max())
print("\nbasic stats:")
print(df.describe())
```

```python
#step 4 - life expectancy over time for all countries
plt.figure(figsize=(12,6))
sns.lineplot( data=df, x="Year", y="Life_Expectancy", hue="Country")
plt.title("Life Expectancy Over Time by Country (1970-2020)")
plt.xlabel("Year")
plt.ylabel("Life Expectancy (years)")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
#step 5 - spending vs life expectancy
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x="Spending_USD",
                y="Life_Expectancy",
                hue="Country",
                size="Year",
                sizes=(20,200))
plt.title("healthcare Spending vs Life Expectancy")
plt.xlabel("healthcare Spending (USD)")
plt.ylabel("Life Expectancy (years)")
plt.tight_layout()
plt.show()
```

```python
# step 6 - average spending per country bar chart
avg_spending = df.groupby("Country")["Spending_USD"].mean().sort_values()

plt.figure(figsize=(10,6))
avg_spending.plot(kind="bar", color=["#00c896","#6c63ff","#f4b942","#ff6b6b","#f472b6","#8899aa"])
plt.title("Average Healthcare Spending by Country(1970-2020)")
plt.xlabel("Country")
plt.ylabel("Average Spending (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
